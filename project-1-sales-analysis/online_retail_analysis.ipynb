## 6. Business Analysis

### Question 1: What is the total sales revenue?

```python
sales_df['Revenue'].sum()
```

### Question 2: Which products generate the highest revenue?

```python
top_products = (
    sales_df.groupby('Description')['Revenue']
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

top_products
```

```python
product_sales = sales_df[
    ~sales_df['Description'].astype(str).str.contains(
        'POSTAGE|DOTCOM POSTAGE',
        case=False,
        na=False
    )
].copy()

top_products = (
    product_sales.groupby('Description')['Revenue']
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

top_products
```

### Question 3: Which countries generate the most revenue?

```python
country_sales = (
    sales_df.groupby('Country')['Revenue']
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

country_sales
```

### Question 4: How does revenue change over time?

```python
sales_df['YearMonth'] = sales_df['InvoiceDate'].dt.to_period('M')

monthly_sales = (
    sales_df.groupby('YearMonth')['Revenue']
    .sum()
)

monthly_sales
```

```python
customer_df = sales_df.dropna(subset=['Customer ID']).copy()

customer_df.shape
```

### Question 5: Which customers generate the most revenue?

```python
top_customers = (
    customer_df.groupby('Customer ID')['Revenue']
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

top_customers
```

```python
product_performance = (
    product_sales.groupby('Description')
    .agg(
        Revenue=('Revenue', 'sum'),
        Units_Sold=('Quantity', 'sum'),
        Transactions=('Invoice', 'count')
    )
    .sort_values('Revenue')
)

product_performance.head(20)
```

### Question 6: Which products have weak sales performance?

```python
weak_products = (
    product_performance[
        product_performance['Transactions'] >= 10
    ]
    .sort_values('Revenue')
    .head(10)
)

weak_products
```

### Question 7: What is the cancellation rate?

```python
cancellations = df_clean[
    df_clean['Invoice'].astype(str).str.startswith('C')
]

cancellations.shape
```

```python
total_invoices = df_clean['Invoice'].nunique()

cancelled_invoices = cancellations['Invoice'].nunique()

cancellation_rate = (cancelled_invoices / total_invoices) * 100

total_invoices, cancelled_invoices, cancellation_rate
```

```python
customer_orders = (
    customer_df.groupby('Customer ID')['Invoice']
    .nunique()
)

total_customers = customer_orders.nunique()

repeat_customers = (customer_orders > 1).sum()

repeat_customer_rate = (repeat_customers / total_customers) * 100

total_customers, repeat_customers, repeat_customer_rate
```

### Question 8: How many customers are repeat customers?

```python
total_customers = customer_orders.index.nunique()

repeat_customers = (customer_orders > 1).sum()

repeat_customer_rate = (repeat_customers / total_customers) * 100

total_customers, repeat_customers, repeat_customer_rate
```
